# NLP Lab: Rule-Based Techniques and Sentiment Lexicons

## Objective
Learn basic sentiment analysis using:
- Rule-Based Approach
- Sentiment Lexicon (Dictionary)

No Machine Learning is used in this lab.


## Theory

A **Rule-Based NLP System** uses manually written rules instead of training a model.

Example Rules:
- Positive words → Positive sentiment
- Negative words → Negative sentiment

A **Sentiment Lexicon** is simply a dictionary containing words with sentiment scores.

Example:

| Word | Score |
|------|------:|
| good | +1 |
| excellent | +2 |
| bad | -1 |
| terrible | -2 |


```
# Step 1: Create a sentiment lexicon

sentiment_lexicon = {
    "good": 1,
    "great": 2,
    "excellent": 3,
    "love": 2,
    "happy": 1,
    "bad": -1,
    "poor": -2,
    "terrible": -3,
    "hate": -2,
    "worst": -3
}

print(sentiment_lexicon)
```

```
# Step 2: Function to calculate sentiment score

def sentiment_score(sentence):
    words = sentence.lower().split()

    score = 0

    for word in words:
        score += sentiment_lexicon.get(word, 0)

    return score
```

```
# Step 3: Test sentences

sentences = [
    "This product is good",
    "I love this phone",
    "This is the worst laptop",
    "The service is terrible",
    "The food is excellent",
    "The movie is bad"
]

for s in sentences:
    print(s)
    print("Score:", sentiment_score(s))
    print("-"*40)
```

```
# Step 4: Classify sentiment

def predict(sentence):
    score = sentiment_score(sentence)

    if score > 0:
        return "Positive 😊"
    elif score < 0:
        return "Negative 😞"
    else:
        return "Neutral 😐"


for s in sentences:
    print(s)
    print("Prediction:", predict(s))
    print()
```

## Real-Life Example

Imagine Flipkart receives customer reviews.

Review:
> "The delivery was excellent and the product is good."

Words found:
- excellent = +3
- good = +1

Total Score = +4

Prediction = Positive


```
# Practice

my_review = input("Enter your review: ")

print("Score:", sentiment_score(my_review))
print("Sentiment:", predict(my_review))
```

# Limitations

Rule-Based NLP cannot understand:

- Negation
  - "not good"

- Sarcasm
  - "Great! Another delayed delivery."

- Context
  - "The battery is light."

Modern NLP models like BERT solve many of these problems.
