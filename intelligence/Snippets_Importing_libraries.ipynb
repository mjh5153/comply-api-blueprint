```
# Ex: find if email is spam or not -- spam ? output class is zero ? non spam ? 1 ? spam

import numpy as np
import matplotlib.pyplot as plt
# create input values -
x = np.arange(-10,11,1)
print(x)
```

```
# Sigmoid function
def sigmoid(x):
  return 1/(1+ np.exp(-x)) # x is valude of zed
y = sigmoid(x)
print(y)

plt.plot(x,y)
plt.title('sigmoid')
plt.grid()
plt.show()
```

```
def tanh(x):
  return np.tanh(x) # if don't pass x, will get error
y = tanh(x)

print(y)

plt.plot(x,y)
plt.title('tanh')
plt.grid()
plt.show()
```

```
#relu Activation -

def relu(x):
  return np.maximum(0,x)

y= relu(x)
print(y)

plt.plot(x,y)
plt.title('relu')
plt.grid()
plt.show()
```

```
# negative values are 0

#softmax function

scores = np.array([2.5, 1.2, 0.3])
softmax = np.exp(scores) / np.sum(np.exp(scores))

print(softmax) # output will be probability - most weighted inpuut has most prob
```
