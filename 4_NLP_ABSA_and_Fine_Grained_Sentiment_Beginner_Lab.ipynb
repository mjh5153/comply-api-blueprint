
## Theory

### Overall Sentiment
One label is assigned to the entire review.

Example:

> "The camera is excellent but the battery is poor."

Overall sentiment may be **Neutral** because it contains both positive and negative opinions.

### Aspect-Based Sentiment Analysis (ABSA)

ABSA identifies the sentiment of each product aspect separately.

Camera → Positive

Battery → Negative


```python
review = "The camera is excellent but the battery is poor."

print("Customer Review:")
print(review)
```

## Step 1 - Define aspects and sentiment words

```python
aspects=["camera","battery","display","performance","price","delivery"]

positive_words=["excellent","good","great","amazing","best","fast","love"]
negative_words=["poor","bad","worst","slow","terrible","hate"]

print(aspects)
```

## Step 2 - Extract aspects from the review

```python
found=[]

text=review.lower()

for aspect in aspects:
    if aspect in text:
        found.append(aspect)

print("Detected Aspects:",found)
```

## Step 3 - Assign sentiment for each detected aspect

```python
results=[]

for aspect in found:

    start=max(0,text.find(aspect)-25)
    end=min(len(text),text.find(aspect)+35)

    window=text[start:end]

    sentiment="Neutral"

    if any(word in window for word in positive_words):
        sentiment="Positive"

    if any(word in window for word in negative_words):
        sentiment="Negative"

    results.append([aspect,sentiment])

df=pd.DataFrame(results,columns=["Aspect","Sentiment"])

df
```


## Fine-Grained Sentiment

Instead of only:

- Positive
- Negative

Some applications use:

- Very Positive
- Positive
- Neutral
- Negative
- Very Negative


```python
reviews=[
"The camera is excellent.",
"The display is good.",
"The battery is poor.",
"The delivery is terrible.",
"The performance is amazing."
]

for r in reviews:

    label="Neutral"

    r_low=r.lower()

    if any(w in r_low for w in ["excellent","amazing","best"]):
        label="Very Positive"
    elif any(w in r_low for w in ["good","love","fast"]):
        label="Positive"
    elif any(w in r_low for w in ["poor","bad"]):
        label="Negative"
    elif any(w in r_low for w in ["terrible","worst","hate"]):
        label="Very Negative"

    print(f"{r} --> {label}")
```
